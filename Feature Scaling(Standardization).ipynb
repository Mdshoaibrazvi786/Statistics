```python
import pandas  as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
df = pd.read_excel("/Users/mr.shoaib/Downloads/Copy of loan.xlsx")
df.head(3)
```

```python
df.isnull().sum()
```

```python
df["ApplicantIncome"].fillna(df["ApplicantIncome"].mean(),inplace=True)
```

```python
sns.displot(df['ApplicantIncome'],bins=70)
plt.show()
```

```python
sns.distplot(df['ApplicantIncome'],bins=40)
plt.show()
```

```python
df.describe()
```

```python
from sklearn.preprocessing import StandardScaler
```

```python
ss = StandardScaler()
ss.fit(df[['ApplicantIncome']])
```

```python
ss.transform(df[['ApplicantIncome']])
```

```python
#Store Dataset
```

```python
df['ApplicantIncome_ss'] = pd.DataFrame(ss.transform(df[['ApplicantIncome']]),columns=["X"])
```

```python
df.head(5)
```

```python
df.describe()
```

```python
sns.distplot(df['ApplicantIncome_ss'],bins=40)
plt.show()
```

```python
plt.figure(figsize=(12,5))
plt.subplot(1,2,1)
plt.title('Before')
sns.distplot(df['ApplicantIncome_ss'])

plt.subplot(1,2,2)
plt.title('After')
sns.displot(df['ApplicantIncome'])

plt.show()
```
