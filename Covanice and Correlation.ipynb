```python
data_corr = af.select_dtypes(["float64","int64"]).corr()
data_corr
```

```python
data_cov = af.select_dtypes(["float64","int64"]).cov()
data_cov
```

```python
sns.heatmap(data_corr,annot=True)
plt.figure(figsize=(4,3))
plt.show()
```

```python
sns.heatmap(data_cov,annot=True)
plt.figure(figsize=(4,3))
plt.show()
```
