```python
import numpy as np
import pandas as  pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
dataset = pd.read_excel("C:/Users/mr.shoaib/Downloads/DataSet.xlsx")
```

```python
dataset
```

```python
md = np.median(dataset["Age"])
```

```python
dataset["Age"].median()
```

```python
mn= np.mean(dataset["Age"])
```

```python
dataset["Age"].mean()
```

```python
sns.histplot(x="Age",data=dataset,bins=[i for i in range(0,81,10)])
plt.plot([mn for i in range(0, 300)],[i for i in  range(0,300)],c="red")
plt.plot([md for i in range(0, 300)],[i for i in range(0,300)],c="blue")
plt.show()
```

```python
mo=dataset["Revenue (USD)"].mode()[0]
```

```python
mo=dataset["Revenue (USD)"].value_counts()
```

```python
sns.histplot(x="Age",data=dataset,bins=[i for i in range(0,81,10)])
plt.plot([mn for i in range(0, 300)],[i for i in  range(0,300)],c="red")
plt.plot([md for i in range(0, 300)],[i for i in range(0,300)],c="blue")
plt.plot([mo for i in range(0, 300)],[i for i in range(0,300)],c="green")
plt.show()
```
